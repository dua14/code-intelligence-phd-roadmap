Day 1: Python Refresher🎯 ObjectiveMaster fast code and data manipulation using Lists, Dictionaries, Generators, and List Comprehensions. These core Python constructs will serve as the engine for parsing, cleaning, and preprocessing code datasets in Phase 1.  📚 Concept Notes1. Lists vs. GeneratorsLists: Store all elements in memory at once. Perfect for quick data operations on small datasets.Generators: Compute values on-the-fly (lazy evaluation). Essential when handling massive source code datasets or parsing thousands of files without crashing your memory (RAM).2. Dictionary & List ComprehensionsList and dictionary comprehensions allow you to filter, transform, and map code tokens or snippets concisely without relying on heavy for loops.💻 Hands-on Code SnippetsHere are practical examples tailored to Code Intelligence workflows:Example A: Extracting & Cleaning Java Tokens (List Comprehension)

```python
# Raw code tokens extracted from a Java method
raw_tokens = ["public", "static", "void", "main(", "String[]", "args)", "{", "", "   "]

# Clean tokens: remove whitespace and empty strings, lowercase keywords
cleaned_tokens = [token.strip().lower() for token in raw_tokens if token.strip()]

print("Cleaned Tokens:", cleaned_tokens)
# Output: ['public', 'static', 'void', 'main(', 'string[]', 'args)', '{']
```

```python
tokens = ["class", "public", "void", "public", "int", "class", "public"]

# Create a frequency count map of keywords
token_counts = {token: tokens.count(token) for token in set(tokens)}

print("Token Frequencies:", token_counts)
# Output: {'void': 1, 'class': 2, 'int': 1, 'public': 3}
```

Example C: Memory-Efficient File Reader (Generator Expression)

```python
def stream_java_lines(code_snippet: str):
    """Yields non-empty Java code lines one by one."""
    for line in code_snippet.strip().split('\n'):
        cleaned = line.strip()
        if cleaned and not cleaned.startswith("//"):  # Skip empty lines and simple comments
            yield cleaned

sample_code = """
public class HelloWorld {
    // This is a comment
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
"""

# Consume generator
for line in stream_java_lines(sample_code):
    print("Streamed Line:", line)
```

Essential OperationsReshaping: Converting 1D token vector sequences into batch shapes or 2D matrix representations.Matrix Operations & Cosine Similarity: Computing dot products and magnitudes to measure how similar two code vectors or embeddings are to each other.💻 Hands-on Code SnippetsHere are practical examples applying NumPy to code vector representations:Example A: Array Creation, Slicing & Reshaping

Example B: Matrix Dot Product & Cosine Similarity

```python
import numpy as np

# Vector representation of Code Snippet A and Code Snippet B
vec_a = np.array([0.2, 0.8, 0.5, 0.1])
vec_b = np.array([0.1, 0.1, 0.1, 0.1])

# Dot Product
dot_product = np.dot(vec_a, vec_b)

# L2 Norm (Magnitude)
norm_a = np.linalg.norm(vec_a)
norm_b = np.linalg.norm(vec_b)

# Cosine Similarity: measures semantic closeness between 2 code vectors
cosine_sim = dot_product / (norm_a * norm_b)

print(f"Dot Product: {dot_product:.4f}")
print(f"Cosine Similarity: {cosine_sim:.4f}")
```

Day 3: Pandas Essentials🎯 ObjectiveMaster DataFrames, Filtering, and Handling Missing Code Data in structured datasets. Pandas will be your primary tool for organizing datasets, cleaning raw code snippets, and managing metadata like repository names, line counts, and method signatures.  📚 Concept Notes1. DataFrames in Code NLPCode datasets are rarely just raw text. They are usually structured tables containing:code_id: Unique identifiercode_snippet: The actual source code (e.g., Java method body)docstring / comment: The corresponding descriptionlanguage & stars: Repository metadata2. Essential OperationsFiltering: Dropping non-Java code, filtering out code snippets that are too short/long, or isolating specific functions.Handling Missing Data (NaN): Replacing or dropping missing comments or empty methods before feeding them to downstream models.💻 Hands-on Code SnippetsHere are practical examples tailored to building and cleaning a Java code dataset:Example A: Creating & Filtering a Code DataFrame

```python
import pandas as pd

# Sample structured raw dataset of Java functions
data = {
    "method_id": [101, 102, 103, 104, 105],
    "code_snippet": [
        "public int add(int a, int b) { return a + b; }",
        "public void log(String msg) { System.out.println(msg); }",
        None,  # Missing code snippet
        "private boolean isValid() { return true; }",
        ""     # Empty string snippet
    ],
    "docstring": [
        "Adds two integers.",
        None,  # Missing comment
        "Placeholder function",
        "Checks validity.",
        "Empty method body"
    ],
    "line_count": [1, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# Filter: Retain only valid (non-null and non-empty) code snippets
cleaned_df = df[df["code_snippet"].notna() & (df["code_snippet"].str.strip() != "")]

print("Filtered DataFrame:")
print(cleaned_df[["method_id", "code_snippet"]])
```

Example B: Handling Missing Values & Adding Derived Features

```python
# Fill missing docstrings with a placeholder
df["docstring"] = df["docstring"].fillna("No documentation provided")

# Create a new feature: Character length of the code snippet
df["char_length"] = df["code_snippet"].apply(lambda x: len(x) if isinstance(x, str) else 0)

print(df[["method_id", "docstring", "char_length"]])
```

Day 4: Data Wrangling🎯 ObjectiveMaster Scraping/Parsing raw text and JSON code snippets into clean datasets. Real-world code datasets (like CodeSearchNet or custom GitHub repos) usually arrive in JSON or JSON Lines (.jsonl) formats containing messy string encodings, escaping issues, and extra metadata.  📚 Concept Notes1. JSON Lines (.jsonl) FormatIn AI for code, huge datasets are saved in .jsonl files where each line is an independent JSON object. This allows processing millions of code functions memory-efficiently without loading the entire JSON file at once.2. Key Data Wrangling Tasks for CodeParsing Strings into JSON: Extracting nested attributes like docstring, func_name, or code.String Sanitization: Normalizing carriage returns (\r\n to \n), stripping trailing whitespace, and unescaping special characters.Flattening: Unpacking nested dictionaries directly into a clean Pandas DataFrame for downstream feature extraction.💻 Hands-on Code SnippetsHere are practical examples of loading raw JSON code entries and converting them into a structured dataset:Example A: Processing Messy JSON Code Entries

```python
import json
import pandas as pd

# Raw JSON string output simulating a stream or .jsonl file response
raw_json_data = """
[
    {"repo": "apache/kafka", "func_name": "sendMessage", "code": "public void sendMessage(String msg) {\\n    System.out.println(msg);\\n}", "docstring": "Sends a message.\\r\\n"},
    {"repo": "spring/framework", "func_name": "checkStatus", "code": "public boolean checkStatus() {\\n    return true;\\n}", "docstring": null}
]
"""

# Step 1: Parse raw JSON string into Python dictionaries
parsed_data = json.loads(raw_json_data)

# Step 2: Wrangle and clean entries
cleaned_records = []
for entry in parsed_data:
    cleaned_records.append({
        "repository": entry.get("repo", "unknown"),
        "function_name": entry.get("func_name"),
        # Clean newlines and carriage returns in the code string
        "code_clean": entry["code"].replace("\r\n", "\n").strip() if entry.get("code") else "",
        # Handle null or missing docstrings gracefully
        "docstring_clean": entry["docstring"].strip() if entry.get("docstring") else "No docstring"
    })

# Step 3: Convert directly into a Pandas DataFrame
df = pd.DataFrame(cleaned_records)
print(df[["repository", "function_name", "docstring_clean"]])
```

Day 5: Probability & Linear Algebra🎯 ObjectiveMaster Dot products, Cosine Similarity, and Vector representations for code. These foundational math concepts allow models to convert source code into high-dimensional vectors (embeddings) and measure semantic similarity between different code snippets or search queries.  📚 Concept Notes1. Code Embeddings as VectorsWhen source code is processed by deep learning models (like CodeBERT or GraphCodeBERT), it is represented as a high-dimensional vector in continuous vector space. Code snippets with similar semantic meaning (e.g., two different functions that sort an array) end up closer together in this vector space.2. Vector Operations for Code SearchDot Product ($A \cdot B$): Measures directional alignment and magnitude combined.Cosine Similarity ($\cos(\theta)$): Measures purely the angle between two vectors regardless of magnitude. It yields a value between $-1$ (opposite) and $1$ (identical). This is the exact math behind semantic code search engines.
  $$\text{Cosine Similarity}(A, B) = \frac{A \cdot B}{\Vert{}A\Vert{} \Vert{}B\Vert{}}$$💻 Hands-on Code SnippetsHere is how to calculate vector similarity manually and with PyTorch tensors:Example A: Manual Cosine Similarity Calculation (Pure Python & Math)

```python
import math

def dot_product(v1, v2):
    return sum(x * y for x, y in zip(v1, v2))

def magnitude(v):
    return math.sqrt(sum(x ** 2 for x in v))

def cosine_similarity(v1, v2):
    dot = dot_product(v1, v2)
    mag_v1 = magnitude(v1)
    mag_v2 = magnitude(v2)
    if mag_v1 == 0 or mag_v2 == 0:
        return 0.0
    return dot / (mag_v1 * mag_v2)

# Vector representations of 3 Java functions
func_add = [0.9, 0.1, 0.2, 0.8]        # Vector for add(a, b)
func_sum = [0.85, 0.15, 0.25, 0.75]    # Vector for calculateSum(a, b) - Semantically similar!
func_delete = [0.1, 0.95, 0.8, 0.05]   # Vector for deleteFile(path) - Unrelated!

print("Similarity (add vs sum):", cosine_similarity(func_add, func_sum))      # High score (~0.99)
print("Similarity (add vs delete):", cosine_similarity(func_add, func_delete)) # Low score (~0.28)
```

Example B: Batch Vector Similarity using PyTorch

```python
import torch
import torch.nn.functional as F

# Query vector: "Search query for sorting an array"
query_vec = torch.tensor([[0.5, 0.8, 0.1, 0.3]])

# Candidate Java function vectors (Batch size: 3, Vector dim: 4)
candidate_funcs = torch.tensor([
    [0.4, 0.9, 0.0, 0.2],   # InsertionSort
    [0.1, 0.0, 0.9, 0.8],   # ConnectToDatabase
    [0.5, 0.7, 0.2, 0.3]    # QuickSort
])

# Compute cosine similarity across the whole batch at once
similarities = F.cosine_similarity(query_vec, candidate_funcs)

print("Similarity Scores for Candidates:", similarities.tolist())
# Output shows higher similarity for sorting functions!
```
