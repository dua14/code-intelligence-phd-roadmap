Text Preprocessing
ObjectiveMaster Lowercasing, Regex tokenization for camelCase code, and Lemmatization.  Source code isn't normal English text. It's full of naming conventions like camelCase (getCalculatedTotal), snake_case (get_calculated_total), punctuation, and language keywords. Day 8 focuses on splitting complex code tokens into sub-words that NLP models can actually learn from.📚 Concept Notes1. Why camelCase Tokenization MattersIn regular NLP, a tokenizer sees getCalculatedTotal as a single unique word. If it has never seen that exact method name before, it fails.By splitting camelCase into sub-words (get, calculated, total), our model can understand the intent of the code even if the full method name is brand new.2. Cleaning Pipelines for CodeRegex Tokenization: Extracting words while splitting on upper-case boundary transitions.Normalization: Lowercasing tokens for classic word-count features (e.g., Bag-of-Words).Lemmatization: Reducing words to their base dictionary form (e.g., calculating $\rightarrow$ calculate).💻 Hands-on Code SnippetsHere is a practical Python script to preprocess and split Java identifier names:Example A: Splitting camelCase and snake_case Java Identifiers

```python
import re

def tokenize_java_identifier(identifier: str) -> list[str]:
    """
    Splits Java camelCase, PascalCase, and snake_case identifiers
    into individual normalized sub-tokens.
    """
    # 1. Replace snake_case underscores with spaces
    cleaned = identifier.replace('_', ' ')

    # 2. Insert space before capital letters in camelCase (e.g. 'getCalculated' -> 'get Calculated')
    cleaned = re.sub(r'([a-z0-9])([A-Z])', r'\1 \2', cleaned)

    # 3. Handle acronyms followed by capital word (e.g. 'XMLParser' -> 'XML Parser')
    cleaned = re.sub(r'([A-Z]+)([A-Z][a-z])', r'\1 \2', cleaned)

    # 4. Tokenize, lower-case, and filter empty strings
    tokens = [token.lower() for token in cleaned.split() if token.strip()]
    return tokens

# Test with common Java identifier styles
test_identifiers = [
    "getUserById",
    "AbstractFactoryBuilder",
    "parse_xml_payload",
    "HTTPResponseHandler"
]

for identifier in test_identifiers:
    print(f"{identifier:25} -> {tokenize_java_identifier(identifier)}")
```

Example B: Lemmatizing Code Docstrings with NLTK

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Download required NLTK resources (run once in Colab)
nltk.download('wordnet', quiet=True)

lemmatizer = WordNetLemmatizer()

docstring_tokens = ["calculating", "retrieved", "indexes", "running"]
lemmatized_tokens = [lemmatizer.lemmatize(token, pos='v') for token in docstring_tokens]

print("Original Tokens:  ", docstring_tokens)
print("Lemmatized (Verbs):", lemmatized_tokens)
# Output: ['calculate', 'retrieve', 'index', 'run']
```
