```python
import torch

print("====================================")
print("DAY 1: BASIC COLAB & GPU CHECK")
print("====================================")

# Check if PyTorch is loaded
print(f"PyTorch Version: {torch.__version__}")

# Check if GPU is ready
if torch.cuda.is_available():
    print(f"✅ GPU is Ready: {torch.cuda.get_device_name(0)}")
else:
    print("⚠️ GPU is NOT enabled! Enable it via Runtime > Change runtime type.")
```

```python
import torch

print("====================================")
print("DAY 2: 20-EPOCH TRAINING LOOP")
print("====================================\n")

# Setup device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# STEP 1: Define Inputs (Study Hours) and Real Targets (Exam Scores)
# Target formula: Score = Hours * 20.0 (e.g., 3 hrs = 60, 5 hrs = 100)
X = torch.tensor([[3.0], [5.0]], device=device)
Y_true = torch.tensor([[60.0], [100.0]], device=device)

# STEP 2: Initial Bad Guess (Weight = 10.0)
W = torch.tensor([[10.0]], device=device, requires_grad=True)
learning_rate = 0.01

print(f"Starting Weight Guess W = {W.item():.2f}\n")
print(f"{'Epoch':<8} | {'Weight (W)':<12} | {'Error (Loss)':<12}")
print("-" * 38)

# STEP 3: 20-Epoch Training Loop
for epoch in range(1, 21):
    # 1. Forward Pass (Make Prediction)
    Y_pred = torch.matmul(X, W)

    # 2. Compute Loss (Mean Squared Error)
    loss = torch.mean((Y_pred - Y_true) ** 2)

    # 3. Backward Pass (AutoGrad computes dL/dW)
    loss.backward()

    # 4. Update Weight & Reset Gradient
    with torch.no_grad():
        W -= learning_rate * W.grad
        W.grad.zero_()  # Clear gradients for the next loop

    # Print progress every 2 epochs
    if epoch % 2 == 0 or epoch == 1:
        print(f"{epoch:<8} | {W.item():<12.4f} | {loss.item():<12.4f}")

print("\n" + "=" * 38)
print(f"🎉 Final Learned Weight W: {W.item():.2f}")
print("Expected perfect weight was 20.00!")
```
